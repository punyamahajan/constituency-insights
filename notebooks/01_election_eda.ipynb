# Constituency election EDA
Light, aggregate-only checks for turnout trend, lagged swing, and feature missingness.

```python
from pathlib import Path
import sqlite3
import matplotlib.pyplot as plt
import pandas as pd
database = Path('../outputs/local/constituency_elections.sqlite')
if not database.is_file():
    raise FileNotFoundError('Build the local database with scripts/build_election_db.py first.')
connection = sqlite3.connect(database)
features = pd.read_sql_query('SELECT * FROM election_features', connection)
```

```python
turnout = pd.read_sql_query('SELECT election_year, AVG(turnout) AS mean_turnout FROM constituency_election_summary GROUP BY election_year ORDER BY election_year', connection)
turnout.plot(x='election_year', y='mean_turnout', marker='o', title='Mean turnout by election year'); plt.ylabel('Turnout (%)');
```

```python
features['swing'].dropna().plot.hist(bins=30, title='Lagged vote-share swing distribution'); plt.xlabel('Percentage points');
```

```python
missingness = features.isna().mean().sort_values(ascending=False).rename('missing_fraction')
missingness.to_frame()
```

```python
connection.close()
```
